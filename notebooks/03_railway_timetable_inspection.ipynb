```python

import pandas as pd
from pathlib import Path

DATA_DIR = Path("../data/raw/railway_timetable")
TIMETABLE_PATH = DATA_DIR / "Train_details_22122017.csv"

timetable = pd.read_csv(
    TIMETABLE_PATH,
    dtype={
        "Train No": "string",
        "Station Code": "string",
        "Source Station": "string",
        "Destination Station": "string",
    },
)

print("Dataset shape:", timetable.shape)
print("\nColumns:")
print(timetable.columns.tolist())

print("\nFirst five rows:")
display(timetable.head())

print("\nData types:")
print(timetable.dtypes)
```

```python

# Check missing values
print("Missing values by column:")
print(timetable.isnull().sum())

# Check train and station coverage
print("\nUnique train numbers:", timetable["Train No"].nunique())
print("Unique station codes:", timetable["Station Code"].nunique())

# Check duplicate train-station sequence records
print(
    "\nDuplicate train/sequence pairs:",
    timetable.duplicated(subset=["Train No", "SEQ"]).sum()
)

# Check station order within each train
timetable["SEQ"] = pd.to_numeric(
    timetable["SEQ"],
    errors="coerce"
)

sequence_check = (
    timetable.sort_values(["Train No", "SEQ"])
    .groupby("Train No")["SEQ"]
    .agg(["count", "min", "max"])
)

print("\nRoute sequence summary:")
display(sequence_check.describe().round(2))

print("\nExample route:")
example_train = timetable["Train No"].dropna().iloc[0]

display(
    timetable[timetable["Train No"] == example_train]
    .sort_values("SEQ")[
        [
            "Train No",
            "Train Name",
            "SEQ",
            "Station Code",
            "Station Name",
            "Arrival time",
            "Departure Time",
            "Distance",
        ]
    ]
)
```
